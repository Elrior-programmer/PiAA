```python
import csv
from collections import defaultdict

data = defaultdict(lambda: {'czas': [], 'mean': [], 'avg': []})

with open('../data/merge_sort_time.csv', 'r') as f:
    reader = csv.reader(f, delimiter=';')
    for row in reader:
        if not row:
            continue
        n = int(row[1])
        data[n]['czas'].append(int(row[2]))
        data[n]['mean'].append(int(row[3]))
        data[n]['avg'].append(float(row[4]))

print(f"{'N':>10} | {'Avg Czas (ns)':>15} | {'Avg Czas (ms)':>15} | {'Mean':>6} | {'Avg':>6}")
print("-" * 65)
for n in sorted(data.keys()):
    avg_czas = sum(data[n]['czas']) / len(data[n]['czas'])
    avg_mean = sum(data[n]['mean']) / len(data[n]['mean'])
    avg_avg  = sum(data[n]['avg'])  / len(data[n]['avg'])
    print(f"{n:>10} | {avg_czas:>15.0f} | {avg_czas/1e6:>15.2f} | {avg_mean:>6.2f} | {avg_avg:>6.2f}")
```
